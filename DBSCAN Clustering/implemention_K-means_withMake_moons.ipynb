```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
X, y = make_blobs(n_samples=500, centers=3, cluster_std=0.60, random_state=42)
```

```python
df = pd.DataFrame(X, columns=['X1', 'X2'])
```

```python
X
```

```python
df
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
inertia = []
K_range = range(1, 11)
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
inertia
```

```python
plt.plot(K_range, inertia, marker='o')
```

```python
kmeans_final = KMeans(n_clusters=3, random_state=42)
```

```python
cluster_labels = kmeans_final.fit_predict(X_scaled)
```

```python
df['Cluster'] = cluster_labels
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='X1', y='X2', hue='Cluster', palette='viridis', s=100)
plt.title('K-Means Clustering Results')
```

Implementing DBSCAN Clustering on the make_moons dataset

```python
from sklearn.datasets import make_moons
```

```python
X_moons, y_moons = make_moons(n_samples=500, noise=0.05, random_state=42)
```

```python
from sklearn.cluster import DBSCAN, KMeans
```

```python
df = pd.DataFrame(X_moons, columns=['X1', 'X2'])
```

```python
#Standardize the data
scaler = StandardScaler()
X_moons_scaled = scaler.fit_transform(df)
```

```python
kmeans_moons = KMeans(n_clusters=2, random_state=42)
kmeans_labels = kmeans_moons.fit_predict(X_moons_scaled)
```

```python
df['KMeans_Cluster'] = kmeans_labels
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='X1', y='X2', hue='KMeans_Cluster', palette='viridis', s=100)
plt.title('K-Means Clustering on make_moons Dataset')
```

```python
dbscan = DBSCAN(eps=0.3, min_samples=5)
dbscan_labels = dbscan.fit_predict(X_moons_scaled)
```

```python
df['DBSCAN_Cluster'] = dbscan_labels
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='X1', y='X2', hue='DBSCAN_Cluster', palette='viridis', s=100)
plt.title('DBSCAN Clustering on make_moons Dataset')
```
