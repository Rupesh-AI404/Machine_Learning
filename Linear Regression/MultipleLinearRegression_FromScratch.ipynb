```python
import numpy as np

# Features: area and bedrooms
X = np.array([
    [800, 2],
    [1000, 2],
    [1200, 3],
    [1500, 3],
    [1800, 4],
    [2000, 4]
], dtype=float)

# Target: house price in millions of NPR
y = np.array([8, 10, 12, 15, 18, 20], dtype=float)
```

```python
print(X.shape)
print(y.shape)
```

```python
X_b = np.column_stack([
    np.ones(X.shape[0]),
    X
])

print(X_b)
```

```python
# Calculate the coefficients using the Normal Equation
beta = np.linalg.lstsq(X_b, y, rcond=None)[0]

print("Intercept:", beta[0])
print("Coefficients:", beta[1:])
```
